# 16 — Physical categorical hierarchies

This notebook presents the frozen-fold extraction, source, quality and waterpoint hierarchy evidence. The accepted model recipe stays fixed, the labelled local test remains unopened, and no oversampling or competition prediction participates.

```python
from pathlib import Path
import pandas as pd

project_dir = next(
    path
    for path in (Path.cwd(), *Path.cwd().parents)
    if (path / 'stage-1-pump-it-up').exists()
)
runtime_dir = project_dir / '.runtime' / 'physical-hierarchy-screen'
families = {
    family: pd.read_csv(runtime_dir / f'{family}-summary.csv', index_col='policy')
    for family in ('extraction', 'source', 'quality', 'waterpoint')
}
hybrids = pd.read_csv(runtime_dir / 'component-hybrids.csv', index_col='hybrid')
print(f"Policies: {sum(len(frame) for frame in families.values())}; refitted challengers: 20")
print(f"Component hybrids: {len(hybrids)}")
```

## Audit boundary

Every hierarchy is deterministic, complete, and covered by the competition frame: extraction 18→13→7 levels, source 10→7→3, quality 8→6, and waterpoint 7→6. This is therefore a granularity and redundancy screen rather than an imputation exercise.

```python
leader_rows = []
for family, frame in families.items():
    leader = frame.iloc[0]
    leader_rows.append({
        'family': family,
        'policy': frame.index[0],
        'mean_accuracy': leader['mean_accuracy'],
        'accuracy_change': leader['accuracy_change'],
        'fold_wins': int(leader['fold_wins']),
        'passes_gate': bool(leader['passes_gate']),
    })
pd.DataFrame(leader_rows).set_index('family').style.format({
    'mean_accuracy': '{:.3%}',
    'accuracy_change': '{:+.3%}',
})
```
