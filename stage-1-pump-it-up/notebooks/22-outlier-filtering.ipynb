## Course-aligned lifecycle

## Interpretation and next loop

Low OOF accuracy identifies difficult rows; it does not prove they are corrupt. Removing unusual pumps deprives the model of evidence needed to predict unusual validation pumps. Exact feature equality likewise does not prove a minority condition label is wrong.

Stop outlier filtering here. The next materially different experiment is one leakage-safe, smoothed multiclass target encoding of predeclared high-cardinality fields. Outer-training rows require inner cross-fitted encodings; outer-validation rows must use mappings fitted on the complete outer-training partition.
