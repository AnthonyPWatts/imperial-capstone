## Course-aligned lifecycle

1. **Define the goal and scope** — add smoothed class evidence without self-encoding.
2. **Gather the data** — reuse frozen development rows and deferred identities.
3. **Explore the data** — audit levels, singleton support, unseen exposure and OOF univariate accuracy.
4. **Clean and preprocess the data** — normalise identity strings and add LGA context.
5. **Select and engineer features** — add three class estimates per selected identity.
6. **Define the machine-learning task** — unchanged three-class classification.
7. **Partition the data** — cross-fit outer-training encodings in five inner folds; fit validation mappings on outer training only.
8. **Select and train candidate methods** — compare location, organisation and combined policies with accepted components.
9. **Evaluate and interpret the results** — apply the gate, component crosses and one bounded contribution curve.
10. **Deploy and iterate** — stop category-wide target encoding and test local spatial outcome features.

## Leakage boundary

`fit_transform` generates each outer-training row's identity values from a separate inner-training partition. `transform` maps outer validation from the complete outer-training partition. Unknown identities receive the training class mean. A row never contributes its own target to its encoded training features.

## Interpretation and next loop

Category-wide class rates are informative but too coarse and suppress repair decisions when used throughout the ensemble. Stop field, smoothing and weight grids.

Next, test cross-fitted local spatial class-rate features. Training rows must use neighbours from an inner-training partition; outer validation must use only outer-training neighbours; invalid coordinates need a prior fallback; and no row may be its own neighbour.
