## Observed ranks and sampling gaps

The sampling-gap map combines the observed locations and output ranks with
the distance to the nearest sampled point. Numbers inside circles are
**observed output ranks: 1 is the highest output**. The background colour
shows distance, as indicated by its colour bar.

In the later plan comparison, **P1, P2 and P3 identify the remaining planned
samples in submission order**. They have not been evaluated and are not
output ranks or predicted outputs.

```python
# Visualisation of input space coverage for query points we have so far 
plot_sampling_gaps(plot_data, save_to=figures / "sampling-gaps.png")
```
